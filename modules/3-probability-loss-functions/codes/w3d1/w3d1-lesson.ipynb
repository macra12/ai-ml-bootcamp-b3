Descriptive Statistics: Mean, Variance, Standard Deviation

```python
# Manual calculation: μ = (1/n) Σ xi
data = [2, 4, 4, 4, 5, 5, 7, 9]
n = len(data)
mean_manual = sum(data) / n  # (2+4+4+4+5+5+7+9) / 8 = 5.0

# NumPy implementation (same formula, vectorized)
import numpy as np
mean_numpy = np.mean(data)   # 5.0
```

Variance

```python
# Step-by-step calculation following the formula
data = np.array([2, 4, 4, 4, 5, 5, 7, 9])
mu = np.mean(data)  # μ = 5.0

# Step 1: (xi - μ) for each point
deviations = data - mu  # [ -3, -1, -1, -1,  0,  0,  2,  4]

# Step 2: (xi - μ)² for each point
squared_deviations = deviations ** 2  # [9, 1, 1, 1, 0, 0, 4, 16]

# Step 3: (1/n) Σ (xi - μ)²
variance_manual = np.sum(squared_deviations) / len(data)  # 32/8 = 4.0

# NumPy implementation (same formula)
variance_numpy = np.var(data)  # 4.0
```

Standard Deviation

```python
# Standard deviation is just the square root of variance
std_manual = np.sqrt(variance_manual)  # √4 = 2.0
std_numpy = np.std(data)               # 2.0
```

Complete Example:

```python
import numpy as np

data = np.array([2, 4, 4, 4, 5, 5, 7, 9])

# All three statistics
mu = np.mean(data)        # 5.0 ← center
var = np.var(data)        # 4.0 ← spread²
sigma = np.std(data)      # 2.0 ← spread

print(f"Mean (μ)       : {mu}")      # 5.0
print(f"Variance (σ²)  : {var}")     # 4.0
print(f"Std Dev (σ)    : {sigma}")   # 2.0

# Standardization (how StandardScaler works)
standardized = (data - mu) / sigma
print(f"After standardization:")
print(f"  New mean: {np.mean(standardized):.10f}")  # ≈ 0 (within float precision)
print(f"  New std : {np.std(standardized):.10f}")   # ≈ 1
```

Bernoulli Distribution: The Coin Flip

```python
import numpy as np

# Bernoulli distribution with p = 0.7
p = 0.7
n_trials = 10000

# Sample from Bernoulli: np.random.binomial(n=1, p=p) gives 0 or 1
outcomes = np.random.binomial(n=1, p=p, size=n_trials)

# Verify the formulas
print(f"Theoretical mean     : μ = p = {p}")
print(f"Simulated mean       : {outcomes.mean():.4f}")  # ≈ 0.7
print(f"Theoretical variance : σ² = p(1-p) = {p * (1-p)}")
print(f"Simulated variance   : {outcomes.var():.4f}")   # ≈ 0.21
```

Uniform Distribution: All Equally Likely

```python
import numpy as np

# Uniform distribution on [2, 8]
a, b = 2, 8
n_samples = 10000

# Sample uniformly: every value equally likely
samples = np.random.uniform(low=a, high=b, size=n_samples)

# Verify the formulas
theoretical_mean = (a + b) / 2       # μ = (2+8)/2 = 5
theoretical_var = (b - a)**2 / 12    # σ² = (8-2)²/12 = 36/12 = 3

print(f"Theoretical mean     : μ = (a+b)/2 = {theoretical_mean}")
print(f"Simulated mean       : {samples.mean():.4f}")  # ≈ 5.0
print(f"Theoretical variance : σ² = (b-a)²/12 = {theoretical_var}")
print(f"Simulated variance   : {samples.var():.4f}")   # ≈ 3.0
```

Normal (Gaussian) Distribution: The Bell Curve

```python
import numpy as np
from scipy import stats

# Normal distribution: μ=100, σ=15 (like IQ scores)
mu, sigma = 100, 15
n_samples = 10000

# Sample from Normal distribution
samples = np.random.normal(loc=mu, scale=sigma, size=n_samples)

# Verify the formulas
print(f"Theoretical mean     : μ = {mu}")
print(f"Simulated mean       : {samples.mean():.4f}")    # ≈ 100
print(f"Theoretical std dev  : σ = {sigma}")
print(f"Simulated std dev    : {samples.std():.4f}")     # ≈ 15

# Verify the 68-95-99.7 rule
within_1sigma = np.sum((samples >= mu - sigma) & (samples <= mu + sigma)) / n_samples
within_2sigma = np.sum((samples >= mu - 2*sigma) & (samples <= mu + 2*sigma)) / n_samples
within_3sigma = np.sum((samples >= mu - 3*sigma) & (samples <= mu + 3*sigma)) / n_samples

print(f"\n68-95-99.7 Rule:")
print(f"  Within μ±1σ: {within_1sigma*100:.1f}% (expected: 68.3%)")
print(f"  Within μ±2σ: {within_2sigma*100:.1f}% (expected: 95.4%)")
print(f"  Within μ±3σ: {within_3sigma*100:.1f}% (expected: 99.7%)")

# Using scipy to calculate the probability density at a point
# This implements the formula: f(x) = (1/σ√2π) * e^(-½((x-μ)/σ)²)
x = 115  # One std dev above mean
pdf_value = stats.norm.pdf(x, loc=mu, scale=sigma)
print(f"\nPDF at x={x}: {pdf_value:.6f}")
```

Understanding the Formula:

```python
# Let's manually calculate the PDF to understand the math
x = 115
manual_pdf = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma)**2)
print(f"Manual PDF calculation: {manual_pdf:.6f}")
# Matches scipy! This is exactly the formula f(x) = (1/σ√2π) * e^(-½((x-μ)/σ)²)
```

Bayes' Theorem: The Update Rule

```python
# Given probabilities
prior_sick = 0.01           # P(sick)
sensitivity = 0.99          # P(positive | sick)
false_pos_rate = 0.05       # P(positive | healthy)

# Step 1: Calculate P(positive) using law of total probability
prob_positive = (sensitivity * prior_sick + 
                 false_pos_rate * (1 - prior_sick))

print(f"P(positive) = {prob_positive:.4f}")  # 0.0594

# Step 2: Apply Bayes' theorem: P(sick | positive)
posterior_sick = (sensitivity * prior_sick) / prob_positive

print(f"P(sick | positive) = {posterior_sick:.4f}")  # 0.1667
print(f"P(sick | positive) = {posterior_sick*100:.1f}%")  # 16.7%
```
